```python
# Encapsulation

# Encapsulation is an object-oriented programming (OOP) principle
# that bundles data (attributes/fields) and the methods (functions)
# that operate on that data into a single unit—typically a class
# while restricting direct access to the internal state from outside the class.(yahi hame sikhna hai)

# private :- it is accessable within the class only
# protected :- it is accessable within the class and during inheritence
# public :- it is access every-where
```

```python
class ATM:
  # data (property)
  def __init__(self,pin,balance):
    self.__pin = pin
    self.__balance = balance
  # method (behaviour)
  def withdraw(self,amount):
    pin = int(input("Enter your pin"))
    if pin != self.pin:
      print("Invalid pin, Try again")
    elif amount <0 or amount > self.balance :
      print(f"Please enter a valid amount, your available balance is {self.balance}")
    else:
      self.balance = self.balance - amount
      print(f"Your new balance is {self.balance}")
    self.menu()

  def deposit(self,amount):
    self.balance = self.balance + amount
    print(f"Your new balance is {self.__balance}")
    self.menu()

  def show_balance(self):
    print(f"Your balance is {self.__balance}")
    self.menu()

  def menu(self):
    print('''
    1 : show_balance
    2 : withdraw
    3 : deposite
    4 : exit''')
    user_input = int(input("Enter your choice"))
    if user_input == 1:
      self.show_balance()
    elif user_input == 2:
      amount = int(input("Enter the amount"))
      self.withdraw(amount)
    elif user_input ==3 :
      amount = int(input("Enter the amount"))
      self.deposit(amount)
    else :
      print("Completed")
```

```python
pintu = ATM(1234,50000)

# pintu is an object of the class ATM
# pintu can access the data and methods of the class ATM

pintu.pin = 566
```
