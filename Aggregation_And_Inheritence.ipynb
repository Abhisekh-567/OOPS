```python
# Inheritance is an object-oriented programming (OOP) mechanism
# where a new class (child / subclass / derived class)
# acquires the properties (attributes) and behaviors (methods)
# of an existing class (parent / superclass / base class).

```

```python
class Animal:  # Parent class
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")

class Dog(Animal):  # Child class inheriting from Animal
    def speak(self):
        print(f"{self.name} barks.")

d = Dog("Buddy")
d.speak()
```

```python
# Aggregation is a specialized form of association in object-oriented programming
# that models a "has-a" relationship where two classes are linked, yet each object has an independent lifecycle.

# If the container (whole) object is destroyed, the contained (part) objects continue to exist independently.
```

```python
class Teacher:
    def __init__(self, name):
        self.name = name


class Department:
    def __init__(self, title, teachers):
        self.title = title
        self.teachers= teachers

    def add_teacher(self, teacher):
        self.teachers.append(teacher)

    def show_teacher(self):
        for teacher in self.teachers:
            print(teacher.name)



t1 = Teacher("Dr. Rao")
t2 = Teacher("Dr. Sharma")
cs_dept = Department("Computer Science", [t1,t2])
t3 = Teacher("Dr. Verma")
cs_dept.add_teacher(t3)

# del cs_dept  # Department is gone, but t1 still exists
# print(t1.name)  # Dr. Rao
cs_dept.show_teacher()
```
